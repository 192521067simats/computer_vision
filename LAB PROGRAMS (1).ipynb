```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Step 1: Upload the image
uploaded = files.upload()

# Get the uploaded image name
filename = next(iter(uploaded))

# Step 2: Read the image
img = cv2.imread(filename)

# Check if image is loaded
if img is None:
    print("Error: Image could not be loaded.")
else:

    # Step 3: Convert image to grayscale
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Step 4: Convert grayscale image to float32
    gray = np.float32(gray)

    # Step 5: Apply Harris Corner Detection
    dst = cv2.cornerHarris(
        gray,
        blockSize=2,
        ksize=3,
        k=0.04
    )

    # Step 6: Dilate the result for better visibility
    dst = cv2.dilate(dst, None)

    # Step 7: Mark detected corners in red
    result = img.copy()
    result[dst > 0.01 * dst.max()] = [0, 0, 255]

    # Step 8: Convert BGR to RGB for displaying with Matplotlib
    original_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    result_rgb = cv2.cvtColor(result, cv2.COLOR_BGR2RGB)

    # Step 9: Display input and output
    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(original_rgb)
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(result_rgb)
    plt.title("Harris Corner Detection")
    plt.axis("off")

    plt.show()

    # Step 10: Save the output
    output_name = "harris_corner_output.jpg"
    cv2.imwrite(output_name, result)

    print("Harris Corner Detection completed successfully.")
    print("Output saved as:", output_name)

    # Step 11: Download the output
    files.download(output_name)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Step 1: Upload the input image
uploaded = files.upload()

# Get uploaded image name
filename = next(iter(uploaded))

# Step 2: Read the image
img = cv2.imread(filename)

# Check image
if img is None:
    print("Error: Image not found!")
else:

    # Step 3: Convert image to grayscale
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Step 4: Apply Sobel filter in X direction
    sobel_x = cv2.Sobel(
        gray,
        cv2.CV_64F,
        1, 0,
        ksize=3
    )

    # Step 5: Apply Sobel filter in Y direction
    sobel_y = cv2.Sobel(
        gray,
        cv2.CV_64F,
        0, 1,
        ksize=3
    )

    # Step 6: Convert results to uint8
    sobel_x = cv2.convertScaleAbs(sobel_x)
    sobel_y = cv2.convertScaleAbs(sobel_y)

    # Step 7: Combine X and Y gradients
    sobel = cv2.addWeighted(
        sobel_x, 0.5,
        sobel_y, 0.5,
        0
    )

    # Step 8: Convert original image for displaying
    original_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Step 9: Display results
    plt.figure(figsize=(15, 5))

    plt.subplot(1, 4, 1)
    plt.imshow(original_rgb)
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 4, 2)
    plt.imshow(sobel_x, cmap="gray")
    plt.title("Sobel X")
    plt.axis("off")

    plt.subplot(1, 4, 3)
    plt.imshow(sobel_y, cmap="gray")
    plt.title("Sobel Y")
    plt.axis("off")

    plt.subplot(1, 4, 4)
    plt.imshow(sobel, cmap="gray")
    plt.title("Sobel Edge Detection")
    plt.axis("off")

    plt.show()

    # Step 10: Save output
    output_name = "sobel_output.jpg"
    cv2.imwrite(output_name, sobel)

    print("Sobel filtering completed successfully.")
    print("Output saved as:", output_name)

    # Step 11: Download output
    files.download(output_name)
```

```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Step 1: Upload image
uploaded = files.upload()

# Get image filename
filename = next(iter(uploaded))

# Step 2: Read image
img = cv2.imread(filename)

if img is None:
    print("Error: Image not found!")
else:

    # Step 3: Convert BGR to RGB for display
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Step 4: Define ROI coordinates
    # Format: image[y1:y2, x1:x2]
    x1, y1 = 100, 100
    x2, y2 = 300, 300

    # Step 5: Crop the ROI
    roi = img[y1:y2, x1:x2]

    # Step 6: Copy the ROI
    copied_roi = roi.copy()

    # Step 7: Paste the copied ROI at a new location
    paste_x = 350
    paste_y = 100

    # Get ROI dimensions
    h, w = copied_roi.shape[:2]

    # Check whether ROI fits inside image
    if paste_y + h <= img.shape[0] and paste_x + w <= img.shape[1]:

        img_pasted = img.copy()

        # Paste ROI
        img_pasted[paste_y:paste_y+h,
                   paste_x:paste_x+w] = copied_roi

        # Convert results to RGB
        roi_rgb = cv2.cvtColor(roi, cv2.COLOR_BGR2RGB)
        pasted_rgb = cv2.cvtColor(img_pasted, cv2.COLOR_BGR2RGB)

        # Step 8: Display results
        plt.figure(figsize=(15, 5))

        plt.subplot(1, 3, 1)
        plt.imshow(img_rgb)
        plt.title("Original Image")
        plt.axis("off")

        plt.subplot(1, 3, 2)
        plt.imshow(roi_rgb)
        plt.title("Cropped ROI")
        plt.axis("off")

        plt.subplot(1, 3, 3)
        plt.imshow(pasted_rgb)
        plt.title("ROI Copied and Pasted")
        plt.axis("off")

        plt.show()

        # Step 9: Save outputs
        cv2.imwrite("cropped_roi.jpg", roi)
        cv2.imwrite("roi_pasted_output.jpg", img_pasted)

        print("ROI cropping completed.")
        print("ROI copying completed.")
        print("ROI pasting completed.")
        print("Files saved successfully.")

        # Step 10: Download outputs
        files.download("cropped_roi.jpg")
        files.download("roi_pasted_output.jpg")

    else:
        print("Error: Paste position is outside the image.")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Step 1: Upload the input image
uploaded = files.upload()

# Get the uploaded image name
filename = next(iter(uploaded))

# Step 2: Read the image
img = cv2.imread(filename)

if img is None:
    print("Error: Image not found!")
else:

    # Step 3: Convert image to grayscale
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Step 4: Create a binary image
    _, binary = cv2.threshold(
        gray, 127, 255, cv2.THRESH_BINARY
    )

    # Step 5: Create a kernel
    kernel = np.ones((5, 5), np.uint8)

    # Step 6: Apply Erosion
    erosion = cv2.erode(
        binary,
        kernel,
        iterations=1
    )

    # Step 7: Display the results
    plt.figure(figsize=(15, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(binary, cmap="gray")
    plt.title("Binary Image")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(erosion, cmap="gray")
    plt.title("Erosion Output")
    plt.axis("off")

    plt.show()

    # Step 8: Save the output
    output_name = "erosion_output.jpg"
    cv2.imwrite(output_name, erosion)

    print("Erosion operation completed successfully.")
    print("Output saved as:", output_name)

    # Step 9: Download the output
    files.download(output_name)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()

# Get uploaded image name
image_name = list(uploaded.keys())[0]

# Read image
img = cv2.imread(image_name)

# Convert BGR to RGB for displaying
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Create a binary image
_, binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

# Create structuring element
kernel = np.ones((5, 5), np.uint8)

# Apply Opening
opening = cv2.morphologyEx(binary, cv2.MORPH_OPEN, kernel)

# Display results
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(binary, cmap="gray")
plt.title("Binary Image")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(opening, cmap="gray")
plt.title("Opening Operation")
plt.axis("off")

plt.show()

# Save output
cv2.imwrite("opening_output.png", opening)

print("Opening operation completed successfully.")
print("Output saved as: opening_output.png")
```
